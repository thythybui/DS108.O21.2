Module chuẩn hóa file address_full_0712.xlsx

1. Dọc dữ liệu

```python
import pandas as pd
import json
import openpyxl

# Load dữ liệu địa chỉ từ file Excel
df = pd.read_excel("D:/LAP TRINH/DS108.O21.2/LAB_5/sources/address_full_0712.xlsx")

# Load file từ điển địa chỉ tổng hợp
with open("vietnam_address.json", "r", encoding="utf-8") as f:
    address_dict = json.load(f)
```

2. Tiền xử lý và chuẩn hóa chuỗi

```python
import re

def normalize_text(text):
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text
```

3. Hàm rút trích thông tin địa chỉ

```python
def extract_address(text, address_dict):
    text = normalize_text(text)
    found_province = ""
    found_district = ""
    found_ward = ""

    for province_data in address_dict.values():
        prov_norm = normalize_text(province_data["name"])
        if prov_norm in text:
            found_province = province_data["name"]

            for district in province_data["district"]:
                dist_norm = normalize_text(f"{district['pre']} {district['name']}")
                if dist_norm in text:
                    found_district = f"{district['pre']} {district['name']}"

                    for ward in district.get("ward", []):
                        ward_norm = normalize_text(f"{ward['pre']} {ward['name']}")
                        if ward_norm in text:
                            found_ward = f"{ward['pre']} {ward['name']}"
                            break
                    break
            break

    clean_text = text
    for part in [normalize_text(found_province), normalize_text(found_district), normalize_text(found_ward)]:
        if part:
            clean_text = clean_text.replace(part, "")
    street = clean_text.strip(",. ")

    return {
        "street": street,
        "ward": found_ward,
        "district": found_district,
        "province": found_province
    }
```

4. Áp dụng vào Excel và chuẩn hoá

```python
components = df["Address"].apply(lambda x: extract_address(x, address_dict)).apply(pd.Series)
df_cleaned = pd.concat([df, components], axis=1)
df_cleaned["normalized_address"] = df_cleaned.apply(
    lambda row: ", ".join(filter(None, [row["street"], row["ward"], row["district"], row["province"]])),
    axis=1
)
```

5. Xuất ra file CSV giống format D_data_address.csv

```python
df_cleaned.to_csv("cleaned_address.csv", index=False)
```
